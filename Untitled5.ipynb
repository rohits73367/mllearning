```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)
```

```python
#box plot
sns.boxplot(x=df['sepal length (cm)'])
plt.show()

#line plot
sns.lineplot(x=df['sepal length (cm)'],y=df['sepal width (cm)'])
plt.show()

#scatter plot
sns.scatterplot(x=df['sepal length (cm)'],y=df['sepal width (cm)'])
plt.show()

#bar plot
sns.barplot(x=df['sepal length (cm)'],y=df['sepal width (cm)'])
plt.show()

#hist plot
sns.histplot(x=df['sepal length (cm)'], y =df['sepal width (cm)'])
plt.show()
```

```python
plt.plot(
    df['sepal length (cm)'],
    df['sepal width (cm)']
)

plt.show()
```

```python
plt.bar(df['sepal length (cm)'], df['sepal width (cm)'])
plt.show()
```

```python
plt.barh(df['sepal length (cm)'], df['sepal width (cm)'])
plt.show()
```

```python
plt.scatter(df['sepal length (cm)'], df['sepal width (cm)'])
plt.show()
```

```python
plt.boxplot(df["sepal length (cm)"])

plt.title("Sepal Length Box Plot")
plt.ylabel("Sepal Length")

plt.show()
```

```python
species_counts = df[
    ["sepal length (cm)", "sepal width (cm)"]
].sum()

plt.pie(
    species_counts,
    labels=species_counts.index,
    autopct="%1.1f%%"
)

plt.title("Sepal Length vs Sepal Width")
plt.show()
```
