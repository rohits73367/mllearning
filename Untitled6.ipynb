# Data Cleaning Summary Report
**Project:** E-Commerce Sales Dataset

## 1. Objective
The objective was to clean, standardize, and validate the raw e-commerce sales dataset to make it suitable for business analysis.

## 2. Dataset Overview

| Description | Before Cleaning | After Cleaning |
|---|---:|---:|
| Rows | 121 | 61 |
| Columns | 14 | 14 |
| Duplicate records | 2 | 0 |
| Duplicate Order IDs | Present | 0 |

*Note: Although the filename mentions 120 rows, the loaded dataset contained 121 rows.*

## 3. Cleaning Operations Performed

- **Order IDs:** Standardized identifiers into the ORD-XXXX format and removed exact duplicate records.
- **Customer Names:** Removed unnecessary spaces and standardized names to Title Case.
- **Emails:** Removed Markdown formatting, converted addresses to lowercase, and replaced malformed addresses with missing values.
- **Phone Numbers:** Removed punctuation and country codes, standardized valid 10-digit numbers, and replaced invalid entries and placeholders with missing values.
- **Categorical Columns:** Standardized Product Category, City, Membership Status, and Payment Method values.
- **Age:** Converted to numeric, replaced impossible ages with missing values, and imputed using the median.
- **Quantity:** Converted written numbers into digits and removed invalid, zero, negative, and extreme values.
- **Unit Price:** Removed currency formatting, converted values to numeric, and treated negative and extreme prices as invalid.
- **Discount:** Standardized percentages into decimal values between 0 and 1. Missing discounts were replaced with 0.
